```python
from IPython import get_ipython
import sys
def run_notebook(path,args=None):
    print(f"\n=== Running {path} ===")
    if args is None:
        args = []

    sys.argv = [path] + args
    get_ipython().run_line_magic("run", path)
    print(f"=== Finished {path} ===\n")

notebooks = [
 "nb1.ipynb",
]

for nb in notebooks:
    try:
        run_notebook(nb,param_dict.get(nb))
    except Exception as e:
        print(f"❌ {nb} failed: {e}")
        # decide if you want to break or continue
        break
```
